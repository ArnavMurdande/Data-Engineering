### CASE STUDY 1 — students attendance analyzer

```python
import pandas as pd

class AttendanceAnalyzer:

    def create_attendance_df(self, data: list) -> pd.DataFrame:

        df = pd.DataFrame(
            data,
            columns=["StudentID", "Department", "Date", "Attendance"]
        )

        return df
```

#### Operation 2 — Calculate Monthly Attendance Rate

```python
def compute_monthly_attendance_rate(self,df: pd.DataFrame) -> pd.DataFrame:

    df = df.copy()

    # Extract YYYY-MM from the Date string
    df["Month"] = df["Date"].str[:7]

    # Count total attendance records
    total = (
        df.groupby(["StudentID", "Month"])
          .size()
          .reset_index(name="Total")
    )

    # Keep only Present records
    present_df = df[df["Attendance"] == "Present"]

    # Count Present records
    present = (
        present_df.groupby(["StudentID", "Month"])
                  .size()
                  .reset_index(name="Present")
    )

    # Combine total and present counts
    result = total.merge(
        present,
        on=["StudentID", "Month"],
        how="left"
    )

    # Students with no Present records get 0
    result["Present"] = result["Present"].fillna(0)

    # Calculate percentage
    result["Attendance Rate"] = (
        result["Present"] / result["Total"] * 100
    )

    return result[
        ["StudentID", "Month", "Attendance Rate"]
    ]
    
```

### Operation 3 — Add Absence Flag

```python
def add_absence_flag(self,df: pd.DataFrame) -> pd.DataFrame:

    df = df.copy()

    df["IsAbsent"] = (
        df["Attendance"] == "Absent"
    ).astype(int)

    return df
```

#### Operation 4 — Students With High Absenteeism

```python
def high_absentees(self,df: pd.DataFrame,threshold: int) -> pd.DataFrame:

    absent = df[df["Attendance"] == "Absent"]

    result = (
        absent.groupby("StudentID")
              .size()
              .reset_index(name="Absence Count")
    )

    result = result[
        result["Absence Count"] > threshold
    ]

    return result
```

#### Operation 5 — Department Attendance Summary

```python
def department_attendance_summary(self,df: pd.DataFrame) -> pd.DataFrame:

    result = pd.crosstab(
        df["Department"],
        df["Attendance"]
    )

    result = result.reindex(
        columns=["Present", "Absent", "Leave"],
        fill_value=0
    )

    result = result.reset_index()

    result.columns.name = None

    return result
```

## CASE STUDY 2 — Word Frequency Counter

```python
import re

def preprocess_text(self, text: str) -> list:

    text = text.lower()

    text = re.sub(
        r'[^\w\s]',
        '',
        text
    )

    words = text.split()

    return words
```

### Operation 2 — Compute Frequency

```python
def compute_word_frequency(self,words: list) -> dict:

    frequency = {}

    for word in words:

        frequency[word] = frequency.get(word, 0) + 1

    return frequency
```

### Operation 3 — Get Most Frequent Word

```python
def get_most_frequent_word(self,freq_dict: dict) -> tuple:

    if not freq_dict:
        return None

    result = max(
        freq_dict.items(),
        key=lambda x: x[1]
    )

    return result
```

### Operation 4 — Filter By Minimum Frequency

```python
def filter_words_by_frequency(self,freq_dict: dict,n: int) -> dict:

    result = {}

    for word, count in freq_dict.items():

        if count >= n:
            result[word] = count

    return result
```

## CASE STUDY 3 — Employee Salary Analyzer

```python
def create_employee_df(self,data: list) -> pd.DataFrame:

    return pd.DataFrame(data,columns=["EmployeeID","Name","Department","Salary"]
    )
```

### Operation 2 — Department Salary Summary

```python
def department_salary_summary(self,df: pd.DataFrame) -> pd.DataFrame:

    result = (
        df.groupby("Department")
          .agg(
              **{
                  "Employee Count": ("EmployeeID", "count"),
                  "Average Salary": ("Salary", "mean")
              }
          )
          .reset_index()
    )

    return result
```

### Operation 3 — Employees Above Department Average

```python
def above_department_average(self,df: pd.DataFrame) -> pd.DataFrame:

    df = df.copy()

    df["Department Average"] = (
        df.groupby("Department")["Salary"]
          .transform("mean")
    )

    result = df[
        df["Salary"] > df["Department Average"]
    ]

    return result[
        ["EmployeeID", "Name", "Department", "Salary"]
    ]
```

### CASE STUDY 4 — Sales Order Analyzer

```python
def create_sales_df(self,data: list) -> pd.DataFrame:

    return pd.DataFrame(data,columns=["OrderID","CustomerID","Product","Category","Quantity","Amount"]
    )
```

### Operation 2 — Category Sales Summary

```python
def category_sales_summary(self,df: pd.DataFrame) -> pd.DataFrame:

    result = (
        df.groupby("Category")
          .agg(
              **{
                  "Total Quantity": ("Quantity", "sum"),
                  "Total Sales": ("Amount", "sum")
              }
          )
          .reset_index()
    )

    return result
```

### Operation 3 — Customer Spending

```python
def customer_spending(self,df: pd.DataFrame) -> pd.DataFrame:

    result = (
        df.groupby("CustomerID")["Amount"]
          .sum()
          .reset_index(name="Total Spending")
    )

    return result
```
